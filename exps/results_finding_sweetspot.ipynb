## Resulting Graph

```python
first_row = lambda df: df.rename(columns={x:y for x,y in zip(
    df.columns,
    range(len(df.columns)))}).iloc[0]

fig, ax = plt.subplots(1, figsize=(5, 15))
plt.xticks(range(4))

ax.plot(first_row(df_lin), label="lin fmnist", color="limegreen")
ax.plot(first_row(df_inv)[:-1], label="inv.pow fmnist", color="darkgreen")
ax.plot(first_row(df_sigm), label="sigm fmnist", color="lightgreen")

ax.plot(first_row(df_lin10), label="lin cifar10", color="blue")
ax.plot(first_row(df_inv10), label="inv.pow cifar10", color="darkblue")
ax.plot(first_row(df_sigm10), label="sigm cifar10", color="lightblue")
    
ax.plot(first_row(df_lin100), label="lin cifar100", color="red")
ax.plot(first_row(df_inv100), label="inv.pow cifar100", color="darkred")
ax.plot(first_row(df_sigm100), label="sigm cifar100", color="pink")

#ax.tick_params(axis='both', which='major', labelsize=fontsize)
ax.set_xticklabels(labels=range(4), rotation=33)
plt.legend()
```

```python
labels = [r"$\hat{\lambda_{"+str(i)+"}}$" for i in range(4)] * 3
labels
```

```python
#fig, ax = plt.subplot




fig, ax = plt.subplots(1, figsize=(10, 10))
plt.xticks(range(12))

for i, row in df_lin10.iterrows():
    #print()
    ax.plot(row, label="linear "+str(i), color=(i/6.+.5, .2, .5-i/6.), marker="*")
    
for i, row in df_inv10.iterrows():
    #print()
    ax.plot(row, label="inv.pow "+str(i), color=(.5, .5-i/8., i/5.+.1,), marker="*")
    
for i, row in df_sigm10.iterrows():
    #print()
    ax.plot(row, label="sigmoid "+str(i), color=(.7-i/5., i/6.+.5, .3), marker="*")

fig.suptitle('Test accuracy on CIFAR10/ResNet18 using Fixed mixup with different lambda factories', fontsize=10)
#ax.tick_params(axis='both', which='major', labelsize=fontsize)
ax.set_xlabel("Lambda vector from steepest to flattest")
ax.set_ylabel("Test accuracy")
ax.set_xticklabels(labels=labels, rotation=0)
fig.text(.18, .15, s=r"$\it{Linear\ factory}$")
fig.text(.48, .15, s=r"$\it{Inv.pow\ factory}$")
fig.text(.78, .15, s=r"$\it{Sigmoid\ factory}$")

lines = [3.5, 7.5]
for xc in lines:
    ax.axvline(x=xc, color='gray', linestyle='--')

ax.axhline(y=85.17, color='red', linestyle='dotted')
fig.text(.5, .745, s=r"$\it{nomix}$", color="red")
fig.tight_layout()
plt.legend(loc="center", bbox_to_anchor=(0, 0, 1, 1), fancybox=True, shadow=True)

plt.savefig("factories_graph.pdf")
```

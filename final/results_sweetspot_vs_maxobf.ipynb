```python
import results_deser as rd
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

#  Relationship between mean test accuracy and N

This notebook produces tables and graphs provided in Subsection 4.7 of the report.

## No-mix

```python
rd.load_testaccs_duplicates("results/r_fmn_cnn_n4_nomix2_0")
```

```python
accs = [rd.load_testaccs("results/r_fmn_cnn_n4_nomix_"+str(i)) for i in range(2)] + \
    [rd.load_testaccs("results/r_fmn_cnn_n4_nomix2_0")]
df_nm_fmn = pd.DataFrame(accs)
df_nm_fmn = df_nm_fmn.transpose()
#df_nm_fmn = pd.DataFrame(df_nm_fmn.values.reshape(1, 4), index=["fmnist"], columns=range(2, 4))
df_nm_fmn.index = ["nomix fmnist"]
df_nm_fmn.columns = range(2, 5)
df_nm_fmn
```

```python
accs = [rd.load_testaccs("results/r_cf10_rn18_n2")] + \
[rd.load_testaccs("results/r_cf10_rn18_n2_"+str(i)) for i in range(1, 3)]
df_nm_cf10 = pd.DataFrame(accs)
df_nm_cf10 = df_nm_cf10.iloc[:, [0]].transpose()
df_nm_cf10.index = ["nomix cifar10"]
df_nm_cf10.columns = range(2, 5)
df_nm_cf10
```

```python
accs = [
    *rd.load_testaccs_duplicates("results/r_cf100_rn18_n4_nomix_0")[1],
    *rd.load_testaccs_duplicates("results/r_cf100_rn18_n4_nomix2_0")[1],
]
df_nm_cf100 = pd.DataFrame(accs)
df_nm_cf100 = pd.DataFrame(df_nm_cf100.values.reshape(1, 3), index=["nomix cifar100"], columns=range(2, 5))
df_nm_cf100
```

```python
df_nomix = pd.concat([df_nm_fmn, df_nm_cf10, df_nm_cf100])
df_nomix
```

```python
df_nomix_wmean = df_nomix.copy()
df_nomix_wmean["mean"] = df_nomix.loc[: , "0":"2"].mean(axis=1).round(2)
df_nomix_wmean
```

## Sweetspot

```python
accs = [rd.load_testaccs("results/r_fmn_cnn_n"+str(i)+"_sweetspot") for i in range(2, 6)]
df_ss_fmn = pd.DataFrame(accs)
df_ss_fmn = pd.DataFrame(np.diag(df_ss_fmn).reshape(1, 4), index=["sweetspot fmnist"], columns=range(2, 6))
df_ss_fmn
```

```python
accs = [rd.load_testaccs("results/r_cf10_rn18_n"+str(i)+"_sweetspot") for i in range(2, 6)]
df_ss_cf10 = pd.DataFrame(accs)
df_ss_cf10 = pd.DataFrame(np.diag(df_ss_cf10).reshape(1, 4), index=["sweetspot cifar10"], columns=range(2, 6))
df_ss_cf10
```

```python
accs = [rd.load_testaccs("results/r_cf100_rn18_n"+str(i)+"_sweetspot") for i in range(2, 6)]
df_ss_cf100 = pd.DataFrame(accs)
df_ss_cf100 = pd.DataFrame(np.diag(df_ss_cf100).reshape(1, 4), index=["sweetspot cifar100"], columns=range(2, 6))
df_ss_cf100
```

```python
df_sweetspot = pd.concat([df_ss_fmn, df_ss_cf10, df_ss_cf100])
df_sweetspot
```

## Maxobf

```python
accs = [rd.load_testaccs("results/r_fmn_cnn_n"+str(i)+"_maxobf") for i in range(2, 6)]
df_mo_fmn = pd.DataFrame(accs)
df_mo_fmn = pd.DataFrame(np.diag(df_mo_fmn).reshape(1, 4), index=["maxobf fmnist"], columns=range(2, 6))
df_mo_fmn
```

```python
accs = [rd.load_testaccs("results/r_cf10_rn18_n"+str(i)+"_maxobf") for i in range(2, 6)]
df_mo_cf10 = pd.DataFrame(accs)
df_mo_cf10 = pd.DataFrame(np.diag(df_mo_cf10).reshape(1, 4), index=["maxobf cifar10"], columns=range(2, 6))
df_mo_cf10
```

```python
accs = [rd.load_testaccs("results/r_cf100_rn18_n"+str(i)+"_maxobf") for i in range(2, 6)]
df_mo_cf100 = pd.DataFrame(accs)
df_mo_cf100 = pd.DataFrame(np.diag(df_mo_cf100).reshape(1, 4), index=["maxobf cifar100"], columns=range(2, 6))
df_mo_cf100
```

```python
df_maxobf = pd.concat([df_mo_fmn, df_mo_cf10, df_mo_cf100])
df_maxobf
```

```python
df_all = pd.concat([df_nomix, df_sweetspot, df_maxobf])
df_all
```

## Graph for maxobf





```python
fig, ax = plt.subplots()
for i, v in df_all.iterrows():
    ax.plot(v, label=v.name)
fig.legend()
```

```python
fig, ax = plt.subplots()
for i, v in df_all.iterrows():
    if not "100" in v.name:
        ax.plot(v, label=v.name)
fig.legend()
```

```python
COLORS = ["#dd0000", "#bc0099", "#008899"]

fig, ax = plt.subplots()
j = 0
for i, v in df_all.iterrows():
    if "fmnist" in v.name:
        ax.plot(v, label=v.name, color=COLORS[j])
        j+=1
ax.set_xlabel("N samples mixed")
ax.set_ylabel("Test accuracy")
ax.axhline(y=89.82, color='r', linestyle='dotted')
fig.legend(loc="lower left", bbox_to_anchor=(.1, .15))
fig.savefig("fmnist_lambda_res.pdf")
```

```python
COLORS = ["#dd0000", "#bc0099", "#008899"]

fig, ax = plt.subplots()
j = 0
for i, v in df_all.iterrows():
    if "cifar10" in v.name and not "cifar100" in v.name:
        ax.plot(v, label=v.name, color=COLORS[j])
        j+=1
ax.set_xlabel("N samples mixed")
ax.set_ylabel("Test accuracy")
ax.axhline(y=85.17, color='r', linestyle='dotted')
fig.legend(loc="lower left", bbox_to_anchor=(.1, .15))
fig.savefig("cifar10_lambda_res.pdf")
```

```python
COLORS = ["#dd0000", "#bc0099", "#008899"]

fig, ax = plt.subplots()
j = 0
for i, v in df_all.iterrows():
    if "cifar100" in v.name:
        ax.plot(v, label=v.name, color=COLORS[j])
        j+=1

ax.axhline(y=54.83, color='r', linestyle='dotted')
ax.set_xlabel("N samples mixed")
ax.set_ylabel("Test accuracy")
fig.legend(loc="lower left", bbox_to_anchor=(.1, .15))
fig.savefig("cifar100_lambda_res.pdf")
```

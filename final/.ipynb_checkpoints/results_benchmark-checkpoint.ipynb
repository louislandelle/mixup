# Benchmark results

This notebook produces the table for the benchmark results provided in Section 3 of the report.

```python
rd.load_testaccs_duplicates("results/r_fmn_cnn_n4_nomix2_0")
```

```python
accs = [rd.load_testaccs("results/r_fmn_cnn_n4_nomix_"+str(i)) for i in range(2)] + \
    [rd.load_testaccs("results/r_fmn_cnn_n4_nomix2_0")]
df_nm_fmn = pd.DataFrame(accs)
df_nm_fmn = df_nm_fmn.transpose()
df_nm_fmn.index = ["nomix fmnist"]
df_nm_fmn.columns = range(3)
```

```python
accs = [rd.load_testaccs("results/r_cf10_rn18_n2")] + \
[rd.load_testaccs("results/r_cf10_rn18_n2_"+str(i)) for i in range(1, 3)]
df_nm_cf10 = pd.DataFrame(accs)
df_nm_cf10 = df_nm_cf10.iloc[:, [0]].transpose()
df_nm_cf10.index = ["nomix cifar10"]
df_nm_cf10.columns = range(3)
```

```python
accs = [
    *rd.load_testaccs_duplicates("results/r_cf100_rn18_n4_nomix_0")[1],
    *rd.load_testaccs_duplicates("results/r_cf100_rn18_n4_nomix2_0")[1],
]
df_nm_cf100 = pd.DataFrame(accs)
df_nm_cf100 = pd.DataFrame(df_nm_cf100.values.reshape(1, 3), index=["nomix cifar100"], columns=range(3))
```

```python
df_nomix = pd.concat([df_nm_fmn, df_nm_cf10, df_nm_cf100])
df_nomix_wmean = df_nomix.copy()
df_nomix_wmean["mean"] = df_nomix.loc[: , "0":"2"].mean(axis=1).round(2)
df_nomix_wmean
```
